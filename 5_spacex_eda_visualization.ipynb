We visualize relationships between features and the landing outcome, then engineer
the feature set (one-hot encoding) used for modeling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style='whitegrid')
```

```python
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv")
print(df.shape); df.head()
```

### Flight Number vs. Payload Mass (colored by outcome)

```python
sns.catplot(y="PayloadMass", x="FlightNumber", hue="Class", data=df, aspect=2)
plt.xlabel("Flight Number", fontsize=15); plt.ylabel("Payload Mass (kg)", fontsize=15); plt.show()
```

### Flight Number vs. Launch Site

```python
sns.catplot(x="FlightNumber", y="LaunchSite", hue="Class", data=df, aspect=2)
plt.xlabel("Flight Number", fontsize=15); plt.ylabel("Launch Site", fontsize=15); plt.show()
```

### Payload Mass vs. Launch Site

```python
sns.catplot(x="PayloadMass", y="LaunchSite", hue="Class", data=df, aspect=2)
plt.xlabel("Payload Mass (kg)", fontsize=15); plt.ylabel("Launch Site", fontsize=15); plt.show()
```

### Success rate of each orbit type

```python
orbit_rate = df.groupby('Orbit')['Class'].mean().sort_values()
orbit_rate.plot(kind='bar', figsize=(10,5))
plt.xlabel("Orbit", fontsize=14); plt.ylabel("Success Rate", fontsize=14); plt.show()
```

### Flight Number vs. Orbit Type

```python
sns.catplot(x="FlightNumber", y="Orbit", hue="Class", data=df, aspect=2)
plt.xlabel("Flight Number", fontsize=15); plt.ylabel("Orbit", fontsize=15); plt.show()
```

### Payload Mass vs. Orbit Type

```python
sns.catplot(x="PayloadMass", y="Orbit", hue="Class", data=df, aspect=2)
plt.xlabel("Payload Mass (kg)", fontsize=15); plt.ylabel("Orbit", fontsize=15); plt.show()
```

### Yearly average success trend

```python
df['Year'] = pd.to_datetime(df['Date']).dt.year
yearly = df.groupby('Year')['Class'].mean()
plt.plot(yearly.index, yearly.values, marker='o')
plt.xlabel("Year", fontsize=14); plt.ylabel("Success Rate", fontsize=14); plt.show()
```

### Feature Engineering
Select the modeling features and one-hot encode the categorical columns
(`Orbit`, `LaunchSite`, `LandingPad`, `Serial`) into the `features_one_hot` matrix.

```python
features = df[['FlightNumber','PayloadMass','Orbit','LaunchSite','Flights','GridFins',
                'Reused','Legs','LandingPad','Block','ReusedCount','Serial']]
features.head()
```

```python
features_one_hot = pd.get_dummies(features,
        columns=['Orbit','LaunchSite','LandingPad','Serial'])
features_one_hot = features_one_hot.astype('float64')
print("Encoded feature matrix shape:", features_one_hot.shape)
features_one_hot.head()
```

```python
features_one_hot.to_csv('dataset_part_3.csv', index=False)
print("Saved dataset_part_3.csv")
```

### Summary
We visualized feature-outcome relationships and built a one-hot encoded feature matrix
`dataset_part_3.csv` for the classification models.
