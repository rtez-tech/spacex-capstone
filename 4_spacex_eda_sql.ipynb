```python
import csv, sqlite3
import pandas as pd

con = sqlite3.connect("my_data1.db")
cur = con.cursor()

%sql sqlite:///my_data1.db
```

### Load the SpaceX dataset into the database

```python
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/Spacex.csv")
df.to_sql("SPACEXTBL", con, if_exists='replace', index=False, method="multi")
```

```python
# Create a clean table that removes blank rows in the Date column
%sql DROP TABLE IF EXISTS SPACEXTABLE;
%sql CREATE TABLE SPACEXTABLE AS SELECT * FROM SPACEXTBL WHERE Date IS NOT NULL
```

### Task 1 — Display the names of the unique launch sites

```python
%sql SELECT DISTINCT Launch_Site FROM SPACEXTABLE;
```

### Task 2 — Display 5 records where launch sites begin with `CCA`

```python
%sql SELECT * FROM SPACEXTABLE WHERE Launch_Site LIKE 'CCA%' LIMIT 5;
```

### Task 3 — Display the total payload mass carried by boosters launched by NASA (CRS)

```python
%sql SELECT SUM(PAYLOAD_MASS__KG_) AS Total_Payload_Mass \
     FROM SPACEXTABLE WHERE Customer = 'NASA (CRS)';
```

### Task 4 — Display the average payload mass carried by booster version F9 v1.1

```python
%sql SELECT AVG(PAYLOAD_MASS__KG_) AS Avg_Payload_Mass \
     FROM SPACEXTABLE WHERE Booster_Version = 'F9 v1.1';
```

### Task 5 — List the date of the first successful landing outcome on a ground pad

```python
%sql SELECT MIN(Date) AS First_Successful_Ground_Landing \
     FROM SPACEXTABLE WHERE Landing_Outcome = 'Success (ground pad)';
```

### Task 6 — List boosters that landed successfully on a drone ship with payload mass between 4000 and 6000 kg

```python
%sql SELECT DISTINCT Booster_Version FROM SPACEXTABLE \
     WHERE Landing_Outcome = 'Success (drone ship)' \
     AND PAYLOAD_MASS__KG_ > 4000 AND PAYLOAD_MASS__KG_ < 6000;
```

### Task 7 — Calculate the total number of successful and failure mission outcomes

```python
%sql SELECT Mission_Outcome, COUNT(*) AS Total \
     FROM SPACEXTABLE GROUP BY Mission_Outcome;
```

### Task 8 — List the booster versions that carried the maximum payload mass (subquery)

```python
%sql SELECT DISTINCT Booster_Version FROM SPACEXTABLE \
     WHERE PAYLOAD_MASS__KG_ = (SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTABLE);
```

### Task 9 — List records (month names, failure landing outcomes, booster versions, launch sites) for drone-ship failures in 2015

```python
%sql SELECT \
        CASE substr(Date,6,2) \
          WHEN '01' THEN 'January' WHEN '02' THEN 'February' WHEN '03' THEN 'March' \
          WHEN '04' THEN 'April'   WHEN '05' THEN 'May'      WHEN '06' THEN 'June' \
          WHEN '07' THEN 'July'    WHEN '08' THEN 'August'   WHEN '09' THEN 'September' \
          WHEN '10' THEN 'October' WHEN '11' THEN 'November' WHEN '12' THEN 'December' \
        END AS Month, \
        Landing_Outcome, Booster_Version, Launch_Site \
     FROM SPACEXTABLE \
     WHERE Landing_Outcome = 'Failure (drone ship)' AND substr(Date,1,4) = '2015';
```

### Task 10 — Rank the count of landing outcomes between 2010-06-04 and 2017-03-20 in descending order

```python
%sql SELECT Landing_Outcome, COUNT(*) AS Outcome_Count \
     FROM SPACEXTABLE \
     WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' \
     GROUP BY Landing_Outcome \
     ORDER BY Outcome_Count DESC;
```
